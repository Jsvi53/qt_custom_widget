## 包络分析过程
包络谱的获取方法
去均值：对原始信号去除直流分量（即去均值），以消除信号中的趋势。
希尔伯特变换：对去均值后的信号进行希尔伯特变换，将其转换为解析信号。
获取包络信号：取解析信号的模（幅值），得到包络信号。
再次去均值：对包络信号再次去均值。
傅里叶变换：对去均值后的包络信号进行快速傅里叶变换（FFT），得到包络谱。

```python
import numpy as np
import matplotlib.pyplot as plt

# 参数设置
fs = 10000  # 采样频率
T = 1       # 采样时间
t = np.arange(0, T, 1/fs)  # 时间向量
A = 1       # 信号幅度
fc = 1000   # 载波频率
fm = 10     # 调制频率

# 生成原始信号
x = A * np.cos(2 * np.pi * fc * t) * np.cos(2 * np.pi * fm * t)

# 绘制原始信号
plt.figure(figsize=(12, 4))
plt.plot(t, x)
plt.title("原始信号")
plt.xlabel("时间 (s)")
plt.ylabel("幅值")
plt.grid()
plt.show()
```

```python
# 去均值
x_mean_removed = x - np.mean(x)

# 绘制去均值后的信号
plt.figure(figsize=(12, 4))
plt.plot(t, x_mean_removed)
plt.title("去均值后的信号")
plt.xlabel("时间 (s)")
plt.ylabel("幅值")
plt.grid()
plt.show()
```

```python
from scipy.signal import hilbert
```

```python
from scipy.signal import hilbert

# 希尔伯特变换
analytic_signal = hilbert(x_mean_removed)

# 绘制解析信号的实部和虚部
plt.figure(figsize=(12, 6))
plt.subplot(2, 1, 1)
plt.plot(t, np.real(analytic_signal))
plt.title("解析信号的实部")
plt.xlabel("时间 (s)")
plt.ylabel("幅值")
plt.grid()

plt.subplot(2, 1, 2)
plt.plot(t, np.imag(analytic_signal))
plt.title("解析信号的虚部")
plt.xlabel("时间 (s)")
plt.ylabel("幅值")
plt.grid()
plt.tight_layout()
plt.show()
```

```python
# 获取包络信号
envelope = np.abs(analytic_signal)

# 绘制包络信号
plt.figure(figsize=(12, 4))
plt.plot(t, envelope)
plt.title("包络信号")
plt.xlabel("时间 (s)")
plt.ylabel("幅值")
plt.grid()
plt.show()
```

```python
# 再次去均值
envelope_mean_removed = envelope - np.mean(envelope)

# 绘制再次去均值后的包络信号
plt.figure(figsize=(12, 4))
plt.plot(t, envelope_mean_removed)
plt.title("再次去均值后的包络信号")
plt.xlabel("时间 (s)")
plt.ylabel("幅值")
plt.grid()
plt.show()
```

```python
# FFT
n = len(envelope_mean_removed)
fft_values = np.fft.fft(envelope_mean_removed)
fft_magnitude = np.abs(fft_values[:n//2])  # 取一半频率范围
frequencies = np.fft.fftfreq(n, 1/fs)[:n//2]  # 频率轴

# 绘制包络谱
plt.figure(figsize=(12, 4))
plt.plot(frequencies, fft_magnitude)
plt.title("包络谱")
plt.xlabel("频率 (Hz)")
plt.ylabel("幅值")
plt.grid()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal

# 定义系统
# 例如：传递函数 H(s) = 1 / (s^2 + 2s + 1)
numerator = [1]
denominator = [1, 2, 1]
system = signal.TransferFunction(numerator, denominator)

# 计算频率响应
frequencies = np.logspace(-2, 2, 500)  # 频率范围从 0.01 到 100 rad/s
w, H = signal.freqresp(system, frequencies)

# 绘制 Nyquist Diagram
plt.figure(figsize=(6, 6))
plt.plot(H.real, H.imag, label='Nyquist Diagram')
plt.plot(H.real, -H.imag, '--', label='Mirror Image')  # 镜像部分
plt.title('Nyquist Diagram')
plt.xlabel('Real Axis')
plt.ylabel('Imaginary Axis')
plt.grid(True)
plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 生成信号
def generate_signal(frequency, sampling_rate, duration):
    t = np.linspace(0, duration, int(sampling_rate * duration), endpoint=False)
    signal = np.sin(2 * np.pi * frequency * t)
    return t, signal

# 应用汉宁窗
def apply_hanning_window(signal):
    window = np.hanning(len(signal))
    windowed_signal = signal * window
    return windowed_signal

# 参数设置
frequency = 5  # 信号频率 (Hz)
sampling_rate = 100  # 采样率 (Hz)
duration = 1  # 信号持续时间 (秒)

# 生成信号
t, signal = generate_signal(frequency, sampling_rate, duration)

# 应用汉宁窗
windowed_signal = apply_hanning_window(signal)

# 绘制原始信号和加窗后的信号
plt.figure(figsize=(12, 6))

plt.subplot(2, 1, 1)
plt.plot(t, signal)
plt.title('Original Signal')
plt.xlabel('Time [s]')
plt.ylabel('Amplitude')

plt.subplot(2, 1, 2)
plt.plot(t, windowed_signal)
plt.title('Signal with Hanning Window')
plt.xlabel('Time [s]')
plt.ylabel('Amplitude')

plt.tight_layout()
plt.show()
```

```python
from PyQt5.QtWidgets import QApplication, QWidget, QPushButton, QVBoxLayout, QLabel
from PyQt5.QtCore import Qt, QPropertyAnimation, QRect, QEasingCurve
import sys

class ZoomTransition(QWidget):
    def __init__(self):
        super().__init__()
        self.resize(800, 600)
        self.setupUI()

    def setupUI(self):
        # 主界面按钮
        self.button = QPushButton("点击放大界面", self)
        self.button.setGeometry(50, 50, 120, 50)
        self.button.clicked.connect(self.zoomInWindow)

    def zoomInWindow(self):
        # 获取按钮在主窗口内的位置和尺寸
        start_rect = self.button.geometry()

        # 创建新界面，设置成主窗口的子控件
        self.new_window = QWidget(self)
        self.new_window.setStyleSheet("background-color: #3498db; border-radius:8px;")
        self.new_window.setGeometry(start_rect)
        
        # 添加一个返回按钮用于缩小回去
        back_btn = QPushButton("返回", self.new_window)
        back_btn.setGeometry(20, 20, 80, 40)
        back_btn.clicked.connect(self.zoomOutWindow)

        # 添加内容示例（可选）
        label = QLabel("这是新界面内容", self.new_window)
        label.setStyleSheet("color:white;font-size:20px;")
        label.adjustSize()
        label.move(50, 100)

        self.new_window.show()

        # 动画放大到主窗口大小
        end_rect = self.rect()

        self.animation = QPropertyAnimation(self.new_window, b"geometry")
        self.animation.setDuration(400)
        self.animation.setStartValue(start_rect)
        self.animation.setEndValue(end_rect)
        self.animation.setEasingCurve(QEasingCurve.OutCubic)
        self.animation.start()

    def zoomOutWindow(self):
        # 缩小动画回到按钮的位置
        end_rect = self.button.geometry()

        self.animation = QPropertyAnimation(self.new_window, b"geometry")
        self.animation.setDuration(400)
        self.animation.setStartValue(self.new_window.geometry())
        self.animation.setEndValue(end_rect)
        self.animation.setEasingCurve(QEasingCurve.InCubic)
        self.animation.finished.connect(self.new_window.deleteLater)
        self.animation.start()

if __name__ == '__main__':
    app = QApplication(sys.argv)
    win = ZoomTransition()
    win.show()
    sys.exit(app.exec_())
```

### 振动数据生成

```python
import numpy as np
import matplotlib.pyplot as plt

fs = 10000  # 采样率
duration = 1.0  # 秒
t = np.linspace(0, duration, int(fs * duration), endpoint=False)

# 基本谐波组成
freqs = [50, 200, 500]        # 可以模拟一些机械共振频率
amps = [1.0, 0.5, 0.2]        # 对应振幅
signal = np.zeros_like(t)

for f, a in zip(freqs, amps):
    signal += a * np.sin(2 * np.pi * f * t)

# 加入高斯白噪声
noise_level = 0.1
noise = np.random.normal(0, noise_level, size=len(t))
signal += noise

# 模拟一个机械冲击（在 0.3 秒附近来一次瞬时冲击衰减）
impact_time = 0.3  # 冲击出现的时刻
impact_index = int(impact_time * fs)
impact_duration = 0.02  # 冲击的持续时间
impact_decay = 0.01     # 冲击的指数衰减系数

for i in range(impact_index, impact_index + int(impact_duration * fs)):
    signal[i] += 2.0 * np.exp(-(i - impact_index) * 1.0 / (impact_decay * fs))

# 画图看看结果
plt.plot(t, signal)
plt.title("模拟机械振动信号示例")
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# --------------------
# 1. 基础参数设置
# --------------------
fs = 20000            # 采样率 (Hz)，这里示例 20 kHz
duration = 1.0        # 信号长度 (秒)
t = np.linspace(0, duration, int(fs * duration), endpoint=False)

# --------------------
# 2. 周期振动成分 (多谐波)
# --------------------
# 例如模拟机械系统中主转轴的基频 100 Hz，及其 2 次、3 次谐波
base_freqs = [100, 200, 300]  # 基频和谐波
amps = [1.0, 0.4, 0.2]        # 每个谐波的幅度，随意指定
signal = np.zeros_like(t)

for f, a in zip(base_freqs, amps):
    signal += a * np.sin(2 * np.pi * f * t)

# --------------------
# 3. 幅度调制 (AM) (可选)
# --------------------
# 假设机械负载略有波动，频率较低 (3 Hz)，影响整体振幅
# AM: amplitude(t) = (1 + 0.2*sin(2*pi*3*t))
am_freq = 3           # 幅值调制的频率 (Hz)
am_depth = 0.2        # 调制深度
am_envelope = 1.0 + am_depth * np.sin(2 * np.pi * am_freq * t)
signal *= am_envelope

# --------------------
# 4. 冲击/碰撞 (脉冲)
# --------------------
# 假设在 0.3 秒、0.6 秒、0.8 秒时发生几次瞬时冲击，随后有阻尼衰减振铃
impact_times = [0.3, 0.6, 0.8]
impact_width = 0.01       # 冲击后的衰减持续时间 (秒)
impact_decay = 0.002      # 指数衰减系数 (越小衰减越快)
impact_amplitude = 2.0    # 冲击脉冲初始幅值

for itime in impact_times:
    start_idx = int(itime * fs)
    end_idx = int((itime + impact_width) * fs)
    for i in range(start_idx, end_idx):
        # 指数衰减的简谐振动形态，这里可随意设计
        signal[i] += impact_amplitude * np.exp(- (i - start_idx) / (impact_decay * fs)) \
            * np.sin(2 * np.pi * 800 * (i - start_idx) / fs)
        # 其中 800 Hz 可理解为结构在冲击后激发的某模态频率

# --------------------
# 5. 叠加噪声
# --------------------
noise_level = 0.2  # 控制噪声强度
noise = np.random.normal(0, noise_level, size=len(t))
signal += noise

# --------------------
# 6. （可选）高通滤波或传感器通道模拟
# --------------------
# 如果要模拟加速度传感器常见的低频截止、高频共振，可使用 scipy.signal
# 高通滤波器或带通滤波器，这里暂不实现，用户可以视需求决定

# --------------------
# 7. 可视化
# --------------------
plt.figure()
plt.plot(t, signal)
plt.title("模拟加速度传感器采集到的振动信号 (时域波形)")
plt.xlabel("Time (s)")
plt.ylabel("Acceleration (a.u.)")
plt.xlim(0, 1.0)
plt.show()

# 也可做个频谱看看
signal_fft = np.fft.rfft(signal)
freqs = np.fft.rfftfreq(len(t), 1 / fs)

plt.figure()
plt.plot(freqs, np.abs(signal_fft))
plt.title("模拟加速度信号的幅度谱")
plt.xlabel("Frequency (Hz)")
plt.ylabel("Amplitude")
plt.xlim(0, 2000)  # 只看低频段
plt.show()
```
