## 包络分析过程
包络谱的获取方法
去均值：对原始信号去除直流分量（即去均值），以消除信号中的趋势。
希尔伯特变换：对去均值后的信号进行希尔伯特变换，将其转换为解析信号。
获取包络信号：取解析信号的模（幅值），得到包络信号。
再次去均值：对包络信号再次去均值。
傅里叶变换：对去均值后的包络信号进行快速傅里叶变换（FFT），得到包络谱。

```python
import numpy as np
import matplotlib.pyplot as plt

# 参数设置
fs = 10000  # 采样频率
T = 1       # 采样时间
t = np.arange(0, T, 1/fs)  # 时间向量
A = 1       # 信号幅度
fc = 1000   # 载波频率
fm = 10     # 调制频率

# 生成原始信号
x = A * np.cos(2 * np.pi * fc * t) * np.cos(2 * np.pi * fm * t)

# 绘制原始信号
plt.figure(figsize=(12, 4))
plt.plot(t, x)
plt.title("原始信号")
plt.xlabel("时间 (s)")
plt.ylabel("幅值")
plt.grid()
plt.show()
```

```python
# 去均值
x_mean_removed = x - np.mean(x)

# 绘制去均值后的信号
plt.figure(figsize=(12, 4))
plt.plot(t, x_mean_removed)
plt.title("去均值后的信号")
plt.xlabel("时间 (s)")
plt.ylabel("幅值")
plt.grid()
plt.show()
```

```python
from scipy.signal import hilbert
```

```python
from scipy.signal import hilbert

# 希尔伯特变换
analytic_signal = hilbert(x_mean_removed)

# 绘制解析信号的实部和虚部
plt.figure(figsize=(12, 6))
plt.subplot(2, 1, 1)
plt.plot(t, np.real(analytic_signal))
plt.title("解析信号的实部")
plt.xlabel("时间 (s)")
plt.ylabel("幅值")
plt.grid()

plt.subplot(2, 1, 2)
plt.plot(t, np.imag(analytic_signal))
plt.title("解析信号的虚部")
plt.xlabel("时间 (s)")
plt.ylabel("幅值")
plt.grid()
plt.tight_layout()
plt.show()
```

```python
# 获取包络信号
envelope = np.abs(analytic_signal)

# 绘制包络信号
plt.figure(figsize=(12, 4))
plt.plot(t, envelope)
plt.title("包络信号")
plt.xlabel("时间 (s)")
plt.ylabel("幅值")
plt.grid()
plt.show()
```

```python
# 再次去均值
envelope_mean_removed = envelope - np.mean(envelope)

# 绘制再次去均值后的包络信号
plt.figure(figsize=(12, 4))
plt.plot(t, envelope_mean_removed)
plt.title("再次去均值后的包络信号")
plt.xlabel("时间 (s)")
plt.ylabel("幅值")
plt.grid()
plt.show()
```

```python
# FFT
n = len(envelope_mean_removed)
fft_values = np.fft.fft(envelope_mean_removed)
fft_magnitude = np.abs(fft_values[:n//2])  # 取一半频率范围
frequencies = np.fft.fftfreq(n, 1/fs)[:n//2]  # 频率轴

# 绘制包络谱
plt.figure(figsize=(12, 4))
plt.plot(frequencies, fft_magnitude)
plt.title("包络谱")
plt.xlabel("频率 (Hz)")
plt.ylabel("幅值")
plt.grid()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal

# 定义系统
# 例如：传递函数 H(s) = 1 / (s^2 + 2s + 1)
numerator = [1]
denominator = [1, 2, 1]
system = signal.TransferFunction(numerator, denominator)

# 计算频率响应
frequencies = np.logspace(-2, 2, 500)  # 频率范围从 0.01 到 100 rad/s
w, mag, phase = signal.bode(system, frequencies)

# 绘制 Bode Plot
plt.figure(figsize=(10, 6))

# 幅频图
plt.subplot(2, 1, 1)
plt.semilogx(w, mag)  # 对数频率轴
plt.title('Bode Plot - Magnitude')
plt.ylabel('Magnitude [dB]')
plt.grid(which='both', axis='both')

# 相频图
plt.subplot(2, 1, 2)
plt.semilogx(w, phase)  # 对数频率轴
plt.title('Bode Plot - Phase')
plt.ylabel('Phase [degrees]')
plt.xlabel('Frequency [rad/s]')
plt.grid(which='both', axis='both')

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal

# 定义系统
# 例如：传递函数 H(s) = 1 / (s^2 + 2s + 1)
numerator = [1]
denominator = [1, 2, 1]
system = signal.TransferFunction(numerator, denominator)

# 计算频率响应
frequencies = np.logspace(-2, 2, 500)  # 频率范围从 0.01 到 100 rad/s
w, H = signal.freqresp(system, frequencies)

# 绘制 Nyquist Diagram
plt.figure(figsize=(6, 6))
plt.plot(H.real, H.imag, label='Nyquist Diagram')
plt.plot(H.real, -H.imag, '--', label='Mirror Image')  # 镜像部分
plt.title('Nyquist Diagram')
plt.xlabel('Real Axis')
plt.ylabel('Imaginary Axis')
plt.grid(True)
plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 生成信号
def generate_signal(frequency, sampling_rate, duration):
    t = np.linspace(0, duration, int(sampling_rate * duration), endpoint=False)
    signal = np.sin(2 * np.pi * frequency * t)
    return t, signal

# 应用汉宁窗
def apply_hanning_window(signal):
    window = np.hanning(len(signal))
    windowed_signal = signal * window
    return windowed_signal

# 参数设置
frequency = 5  # 信号频率 (Hz)
sampling_rate = 100  # 采样率 (Hz)
duration = 1  # 信号持续时间 (秒)

# 生成信号
t, signal = generate_signal(frequency, sampling_rate, duration)

# 应用汉宁窗
windowed_signal = apply_hanning_window(signal)

# 绘制原始信号和加窗后的信号
plt.figure(figsize=(12, 6))

plt.subplot(2, 1, 1)
plt.plot(t, signal)
plt.title('Original Signal')
plt.xlabel('Time [s]')
plt.ylabel('Amplitude')

plt.subplot(2, 1, 2)
plt.plot(t, windowed_signal)
plt.title('Signal with Hanning Window')
plt.xlabel('Time [s]')
plt.ylabel('Amplitude')

plt.tight_layout()
plt.show()
```

```python
from PyQt5.QtWidgets import QApplication, QWidget, QPushButton, QVBoxLayout, QLabel
from PyQt5.QtCore import Qt, QPropertyAnimation, QRect, QEasingCurve
import sys

class ZoomTransition(QWidget):
    def __init__(self):
        super().__init__()
        self.resize(800, 600)
        self.setupUI()

    def setupUI(self):
        # 主界面按钮
        self.button = QPushButton("点击放大界面", self)
        self.button.setGeometry(50, 50, 120, 50)
        self.button.clicked.connect(self.zoomInWindow)

    def zoomInWindow(self):
        # 获取按钮在主窗口内的位置和尺寸
        start_rect = self.button.geometry()

        # 创建新界面，设置成主窗口的子控件
        self.new_window = QWidget(self)
        self.new_window.setStyleSheet("background-color: #3498db; border-radius:8px;")
        self.new_window.setGeometry(start_rect)
        
        # 添加一个返回按钮用于缩小回去
        back_btn = QPushButton("返回", self.new_window)
        back_btn.setGeometry(20, 20, 80, 40)
        back_btn.clicked.connect(self.zoomOutWindow)

        # 添加内容示例（可选）
        label = QLabel("这是新界面内容", self.new_window)
        label.setStyleSheet("color:white;font-size:20px;")
        label.adjustSize()
        label.move(50, 100)

        self.new_window.show()

        # 动画放大到主窗口大小
        end_rect = self.rect()

        self.animation = QPropertyAnimation(self.new_window, b"geometry")
        self.animation.setDuration(400)
        self.animation.setStartValue(start_rect)
        self.animation.setEndValue(end_rect)
        self.animation.setEasingCurve(QEasingCurve.OutCubic)
        self.animation.start()

    def zoomOutWindow(self):
        # 缩小动画回到按钮的位置
        end_rect = self.button.geometry()

        self.animation = QPropertyAnimation(self.new_window, b"geometry")
        self.animation.setDuration(400)
        self.animation.setStartValue(self.new_window.geometry())
        self.animation.setEndValue(end_rect)
        self.animation.setEasingCurve(QEasingCurve.InCubic)
        self.animation.finished.connect(self.new_window.deleteLater)
        self.animation.start()

if __name__ == '__main__':
    app = QApplication(sys.argv)
    win = ZoomTransition()
    win.show()
    sys.exit(app.exec_())
```

```python
from PyQt5.QtWidgets import (QApplication, QWidget, QPushButton, QLabel,
                             QGraphicsOpacityEffect, QGraphicsDropShadowEffect)
from PyQt5.QtCore import (Qt, QPropertyAnimation, QRect, QEasingCurve, pyqtProperty)
import sys

class ZoomTransition(QWidget):
    def __init__(self):
        super().__init__()
        self.resize(800, 600)
        self.setupUI()
        self._radius = 15  # 圆角初始值

    def setupUI(self):
        self.button = QPushButton("点击放大界面", self)
        self.button.setGeometry(50, 50, 120, 50)
        self.button.clicked.connect(self.zoomInWindow)

    # 圆角属性定义
    def getRadius(self):
        return self._radius

    def setRadiusValue(self, value):
        self._radius = value
        self.new_window.setStyleSheet(
            f"background-color: #3498db; border-radius:{value}px;"
        )

    radius = pyqtProperty(int, fget=getRadius, fset=setRadiusValue)

    def zoomInWindow(self):
        start_rect = self.button.geometry()
        end_rect = self.rect()

        self.new_window = QWidget(self)
        self.new_window.setGeometry(start_rect)
        self.setRadiusValue(15)

        # 阴影效果
        shadow = QGraphicsDropShadowEffect()
        shadow.setBlurRadius(25)
        shadow.setOffset(0, 10)
        shadow.setColor(Qt.black)
        self.new_window.setGraphicsEffect(shadow)

        # 透明度效果
        opacity_effect = QGraphicsOpacityEffect(self.new_window)
        self.new_window.setGraphicsEffect(opacity_effect)

        self.new_window.show()

        # 尺寸位置动画
        self.anim_geo = QPropertyAnimation(self.new_window, b"geometry")
        self.anim_geo.setDuration(500)
        self.anim_geo.setStartValue(start_rect)
        self.anim_geo.setEndValue(end_rect)
        self.anim_geo.setEasingCurve(QEasingCurve.InOutCubic)
        self.anim_geo.start()

        # 透明度动画
        self.anim_opacity = QPropertyAnimation(opacity_effect, b"opacity")
        self.anim_opacity.setDuration(500)
        self.anim_opacity.setStartValue(0)
        self.anim_opacity.setEndValue(1)
        self.anim_opacity.setEasingCurve(QEasingCurve.InOutQuad)
        self.anim_opacity.start()

        # 圆角动画
        self.anim_radius = QPropertyAnimation(self, b"radius")
        self.anim_radius.setDuration(500)
        self.anim_radius.setStartValue(15)
        self.anim_radius.setEndValue(0)
        self.anim_radius.setEasingCurve(QEasingCurve.InOutQuad)
        self.anim_radius.start()

        # 加载新界面内容
        label = QLabel("精致的缩放界面切换动画", self.new_window)
        label.setStyleSheet("color:white;font-size:22px;")
        label.adjustSize()
        label.move(40, 80)

        back_btn = QPushButton("返回", self.new_window)
        back_btn.setGeometry(20, 20, 80, 40)
        back_btn.clicked.connect(self.zoomOutWindow)

    def zoomOutWindow(self):
        end_rect = self.button.geometry()

        self.anim_geo.setDirection(QPropertyAnimation.Backward)
        self.anim_geo.start()

        self.anim_opacity.setDirection(QPropertyAnimation.Backward)
        self.anim_opacity.start()

        self.anim_radius.setDirection(QPropertyAnimation.Backward)
        self.anim_radius.start()

        self.anim_geo.finished.connect(self.new_window.deleteLater)

if __name__ == '__main__':
    app = QApplication(sys.argv)
    win = ZoomTransition()
    win.show()
    sys.exit(app.exec_())
```

```python
import sys
from PyQt5.QtCore import Qt, QTimer
from PyQt5.QtGui import QVector3D
from PyQt5.QtWidgets import QApplication, QMainWindow, QVBoxLayout, QWidget
from PyQt5.Qt3DCore import QEntity, QTransform
from PyQt5.Qt3DExtras import Qt3DWindow, QOrbitCameraController, QPhongMaterial, QCuboidMesh
from PyQt5.Qt3DRender import QCamera

class My3DWindow(Qt3DWindow):
    def __init__(self):
        super().__init__()

        # 创建窗口容器
        self.container = QWidget.createWindowContainer(self)
        self.container.setGeometry(0, 0, 800, 600)

        # 初始化相机
        self.camera = self.camera()
        self.camera.setPosition(QVector3D(0, 0, 20))
        self.camera.setViewCenter(QVector3D(0, 0, 0))
        self.camera.setUpVector(QVector3D(0, 1, 0))
        self.camera.setFieldOfView(45)
        self.camera.setAspectRatio(800 / 600)  # 根据窗口尺寸设置宽高比

        # 创建场景根实体
        self.root_entity = QEntity()
        self.create_scene()

        # 相机控制器
        self.cam_controller = QOrbitCameraController(self.root_entity)
        self.cam_controller.setCamera(self.camera)

        # 定时器实现自动缩放
        self.timer = QTimer()
        self.timer.timeout.connect(self.update_camera_position)
        self.timer.start(50)

        # 设置场景根实体
        self.setRootEntity(self.root_entity)

    def create_scene(self):
        # 创建立方体实体
        cube_entity = QEntity(self.root_entity)
        cube_mesh = QCuboidMesh()
        cube_mesh.setXExtent(2.0)
        cube_mesh.setYExtent(2.0)
        cube_mesh.setZExtent(2.0)

        # 材质
        material = QPhongMaterial()
        material.setDiffuse(Qt.red)

        # 变换
        transform = QTransform()

        cube_entity.addComponent(cube_mesh)
        cube_entity.addComponent(material)
        cube_entity.addComponent(transform)

    def update_camera_position(self):
        current_pos = self.camera.position()
        new_z = current_pos.z() - 0.05
        if new_z > 1.0:  # 防止相机过于接近物体
            self.camera.setPosition(QVector3D(current_pos.x(), current_pos.y(), new_z))

class MainWindow(QMainWindow):
    def __init__(self):
        super().__init__()
        self.setWindowTitle("PyQt5 3D Cube Demo")
        self.view = My3DWindow()
        layout = QVBoxLayout()
        layout.addWidget(self.view.container)
        central_widget = QWidget()
        central_widget.setLayout(layout)
        self.setCentralWidget(central_widget)

if __name__ == '__main__':
    app = QApplication(sys.argv)
    window = MainWindow()
    window.resize(800, 600)
    window.show()
    sys.exit(app.exec_())
```

```python
import sys
import numpy as np
import pyqtgraph as pg
import pyqtgraph.opengl as gl
from PyQt5.QtWidgets import QApplication, QMainWindow, QVBoxLayout, QWidget

class ThreeDPlotGL(QMainWindow):
    def __init__(self):
        super().__init__()
        
        # 创建主部件和布局
        main_widget = QWidget()
        self.setCentralWidget(main_widget)
        layout = QVBoxLayout(main_widget)
        
        # 创建OpenGL 3D视图
        self.view = gl.GLViewWidget()
        layout.addWidget(self.view)
        
        # 设置3D场景
        self.view.setCameraPosition(distance=20, elevation=20, azimuth=-35)
        
        # 添加3D坐标轴
        axis = gl.GLAxisItem()
        axis.setSize(10, 10, 10)
        self.view.addItem(axis)
        
        # 生成示例数据
        x = np.linspace(-5, 5, 50)
        y = np.linspace(-5, 5, 50)
        x, y = np.meshgrid(x, y)
        z = np.sin(np.sqrt(x**2 + y**2))
        
        # 创建网格数据项
        colors = np.ones((50, 50, 4)) * (0, 0.5, 1, 1)  # RGBA
        colors[..., 3] = 0.3  # 设置透明度
        self.surface = gl.GLSurfacePlotItem(x=x, y=y, z=z, colors=colors)
        self.surface.shader()['lighting'] = True
        self.view.addItem(self.surface)

if __name__ == '__main__':
    app = QApplication(sys.argv)
    window = ThreeDPlotGL()
    window.setWindowTitle('3D Plot with PyQtGraph')
    window.resize(800, 600)
    window.show()
    sys.exit(app.exec_())
```
